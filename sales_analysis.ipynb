```python
import pandas as pd
from pathlib import Path

from src.analysis import clean_sales_data, get_monthly_revenue, get_top_products

df = pd.read_csv(Path('..') / 'data' / 'sales_data.csv')
clean_df = clean_sales_data(df)
monthly = get_monthly_revenue(clean_df)
top_products = get_top_products(clean_df, 10)

print(monthly.head())
print(top_products.head())

monthly.plot(x='Month', y='Revenue', kind='line', title='Monthly Revenue')
top_products.plot(x='Product', y='Revenue', kind='barh', title='Top Products by Revenue')
```
